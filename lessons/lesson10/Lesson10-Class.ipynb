Cancelled Invoices

"Look at each InvoiceNo, convert it to text and uppercase, then check whether it starts with C. Store the resulting True/False values as cancelled_mask."

For the UCI Online Retail dataset, this is useful because invoice numbers beginning with C indicate cancellations/credit transactions.

```python
cancelled_invoice= df['InvoiceNo'].astype(str).str.upper().str.startswith('C')
cancelled_count = cancelled_invoice.sum()
cancelled_count
```

```python
df[cancelled_invoice].head(20)
```

Loop vs Vectorized Operation

Both approaches can express the same logic. Vectorized pandas operations are generally preferable for straightforward column-wise tabular calculations.

```python
negative_quantity_loop= 0
for quantity in df['Quantity']:
    if quantity < 0:
        negative_quantity_loop += 1
negative_quantity_loop
```

```python
negative_quantity_vect= (df['Quantity']< 0).sum()
negative_quantity_vect
```

Create Sales

```python
df['Sales']= df['Quantity'] * df['UnitPrice']
df.head(10)
```

Date Processing

convert a date column into a proper datetime format and then extract useful parts of the date.

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["Date"]= df["InvoiceDate"].dt.date
df["Year"]= df["InvoiceDate"].dt.year
df['Month']=df["InvoiceDate"].dt.month
df[["InvoiceDate", "Date", "Year", "Month"]].head(20)
```

Merge Activity

This code creates a lookup table that assigns each country to either UK or International

```python
country_lookup= pd.DataFrame({"Country": df["Country"].dropna().unique()})
country_lookup["Region"]=np.where(country_lookup["Country"].eq("United Kingdom"), "UK", "International")
country_lookup.head(20)
```

This code joins the country_lookup table back to the original df DataFrame so that every transaction gets its corresponding Region.

```python
merged_df= df.merge(country_lookup, on="Country", how="left")
merged_df[['Country','Region']].head(10)
```
