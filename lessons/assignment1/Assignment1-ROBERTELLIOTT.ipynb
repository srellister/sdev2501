For 2 numerical columns, compute:
Mean, median, mode, range, variance, standard deviation, quartiles(Q1, Q3), IQR

```python
column_name = "Maximum Temperature (C)"

print(f"The mean for the column {column_name} is: {df[column_name].mean()}")
print(f"The median for the column {column_name} is: {df[column_name].median()}")
print(f"The mode for the column {column_name} is: {df[column_name].mode()}")
df_min = df[column_name].min()
df_max = df[column_name].max()
print(f"The range for the column {column_name} is {df_max-df_min}")
print(f"The variance for the column {column_name} is: {df[column_name].var()}")
print(f"The standard deviation for the column {column_name} is: {df[column_name].std()}")
df_q1 = df[column_name].quantile(0.25)
df_q3 = df[column_name].quantile(0.75)
print(f"The Q1 for the column {column_name} is: {df_q1}")
print(f"The Q3 for the column {column_name} is: {df_q3}")
print(f"The IQR for the column {column_name} is: {df_q3-df_q1}")

```
