For 2 numerical columns, compute:

Compute z-score for one specific data point
the data point we're computing the z-score for is the Maximum Temperature (C) for row 2721420210511: 17.3

```python
column_name = "Maximum Temperature (C)"

x = 17.3
df_mean = df[column_name].mean()
df_std = df[column_name].std()
z_score = (x - df_mean) / df_std
print(f"Z-score for {x} in {column_name} is: {z_score}")
```

Identify outliers

```python
column_name = "Maximum Temperature (C)"
df_q1 = df[column_name].quantile(0.25)
df_q3 = df[column_name].quantile(0.75)
df_iqr = df_q3 - df_q1
lower_bound = df_q1 - (1.5 * df_iqr)
upper_bound = df_q3 + (1.5 * df_iqr)

outliers = df[(df[column_name] < lower_bound) | (df[column_name] > upper_bound)]
print(f"{df_q1} {df_q3}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")
print(f"Number of Outliers: {len(outliers)}")
print(outliers[[column_name]])
```
